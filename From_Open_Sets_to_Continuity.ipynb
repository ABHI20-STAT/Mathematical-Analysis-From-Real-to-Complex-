# **Mathematical Perspective of the Plots (Topological Intuition)**

We are visualizing a **topological idea** using the familiar geometry of the plane $\mathbb{R}^2$. The pictures do **not** claim that every topology looks like circles in the plane; rather, they give the standard intuition that **open sets are built from neighborhoods**.

## **1. The Underlying Set $X$**

In the plots, we first fix a **finite set of points**

$$
X=\{x_1,x_2,\dots,x_6\}\subset \mathbb{R}^2.
$$

Each $x_i$ is a point in the plane, plotted and labeled.  

>**Important Viewpoint:**  
We are treating $X$ as an “abstract set,” but we *embed it* into $\mathbb{R}^2$ so we can draw it.

## **2. Open Balls as Neighborhoods**

In $\mathbb{R}^2$ with the Euclidean distance $d$, an **open ball** centered at $a\in \mathbb{R}^2$ with radius $r>0$ is

$$
B(a,r)=\{y\in \mathbb{R}^2:\ d(y,a)<r\}.
$$

> Geometrically, this is the interior of a circle (boundary not included).  
In the code, each “ball” is drawn as a circle with a lightly shaded interior, representing $B(a,r)$.


## **3. What the Picture is Encoding: Open Sets from Neighborhoods**

The plots define two sets, $U$ and $V$, each formed by taking a **union of open balls**.

### **The Set $U$**
The **left frame** shows:

$$
U = B(a_1,r_1)\ \cup\ B(a_2,r_2).
$$

So $U$ is open in $\mathbb{R}^2$ because it is a union of open sets.

### **The set $V$**
The **right frame** shows:

$$
V = B(b_1,s_1)\ \cup\ B(b_2,s_2).
$$

Similarly, $V$ is open in $\mathbb{R}^2$.

**Main Topological Message:**
> In metric spaces, open sets can be constructed as unions of open balls (neighborhoods).



## **4. Connecting to the Topology $\tau$**

Let $(\mathbb{R}^2,\tau)$ be the plane with its **standard topology**, generated by open balls.

Then:
- $(U \in \tau)$,
- $(V \in \tau)$,

because both are unions of open balls.

The pictures illustrate the “**axioms of topology**” in the background:
1. $\emptyset$ and $\mathbb{R}^2$ are open.
2. Any union of open sets is open.
3. Any finite intersection of open sets is open.



## **5. Where the Finite Set $X$ Fits In**

Although $U$ and $V$ live in $\mathbb{R}^2$, we are interested in how they relate to the finite set $X$.

A natural topological structure on \(X\) comes from the **subspace topology**.

### Subspace topology on \(X\)
Define
$$
\tau_X=\{U\cap X : U\in \tau\}.
$$
Then $(X,\tau_X)$ is a topological space.

> **In words:**
> A subset $A\subseteq X$ is open in $X$ iff there exists an open set $O\subseteq \mathbb{R}^2$ such that  
> $$
 A = O\cap X.
  $$

So, from the plots:
- The open set in the topological space $(X,\tau_X)$ corresponding to $U$ is $U \cap X$.
- The open set corresponding to $V$ is $V \cap X$.

> **Interpretation:**
> - Points of $X$ that lie inside the shaded regions belong to $U \cap X$. (or $V \cap X$.).
> - Points outside do not.

## **6. Neighborhood Interpretation (Key Intuition)**

A point $x\in X$ is “interior” to an open region (like $U$ ) if:

$$
\exists r>0\ \text{such that}\ B(x,r)\subseteq U.
$$

This is the classical neighborhood definition of openness:

> A set is open if every point in it has a neighborhood contained in the set.

Our pictures demonstrate this idea visually: each open ball is literally drawn as a neighborhood region.


## **7. Why Two Frames?**

The two-frame figure is a didactic separation which means that $U,V$ both built from neighborhoods.


This reinforces the principle:

$$
U,\ V\ \text{are open} \quad \Longrightarrow \quad U\cup V\ \text{is open, and } U\cap V\ \text{is open.}
$$

> (If you want, we can add a third frame showing $U\cup V \; \text{or}\; U\cap V$ explicitly.)

## **Summary**

- We fixed a finite set $X\subset \mathbb{R}^2$.
- We constructed two open sets $ (U,V) \in \mathbb{R}^2$ as unions of open balls.
- This illustrates how the standard topology on $\mathbb{R}^2$ is generated by neighborhoods.
- Restricting to $X$  via the subspace topology gives open sets $U \cap X$ and $V \cap X$, which are the genuinely “topological” open sets in $(X,\tau_X)$.

## **Explanation of the Diagram Rotation on $\mathbb{T}$**
The figure has two panels, each representing the circle
$$
\mathbb{T}=\mathbb{R}/(2\pi\mathbb{Z}),
$$
viewed as the unit circle in the plane.

### **Left Panel: Rational Rotation ($\theta=\pi/2$)**
Starting from a point $[\varphi]\in\mathbb{T}$ (marked on the circle), we apply the rotation

$$
R_{\pi/2}([\varphi])=[\varphi+\tfrac{\pi}{2}].
$$

The arrows show successive iterates:
$$
[\varphi]\;\mapsto\;[\varphi+\tfrac{\pi}{2}]\;\mapsto\;[\varphi+\pi]\;\mapsto\;[\varphi+\tfrac{3\pi}{2}]
\;\mapsto\;[\varphi].
$$

After four steps we return exactly to the starting point, i.e. $R_{\pi/2}^4=\mathrm{id}$ on $\mathbb{T}$, and the four marked points form a 4-cycle (a square inscribed in the circle). This visually encodes that every point has
minimal period $4$ under $R_{\pi/2}$.

### **Right Panel: Irrational Rotation ($\theta/2\pi\notin\mathbb{Q}$)**
The orange point marks the starting position. The blue dots represent successive iterates

$$
[\varphi],\ [\varphi+\theta],\ [\varphi+2\theta],\ldots,
$$

and the arrows indicate the first few transitions between consecutive iterates. Because $\theta/2\pi$ is irrational, no iterate returns exactly to the start: there is no $p\ge 1$ with $R_\theta^p([\varphi])=[\varphi]$. Hence the orbit
never forms a finite cycle; instead it keeps producing new points around the circle (and, in fact, such orbits are
dense in $\mathbb{T}$).



# **Mathematical Context of the Diagram**

## **1. The Dynamical System Being Visualized**
Fix a complex parameter $c\in\mathbb{C}$ and consider the quadratic map

$$
f_c:\mathbb{C}\to\mathbb{C},\qquad f_c(z)=z^2+c.
$$

The **discrete-time dynamical system** generated by $f_c$ is obtained by iterating:

$$
z_{n+1}=f_c(z_n)=z_n^2+c,\qquad n=0,1,2,\dots,
$$

starting from a seed (initial condition) $z_0\in\mathbb{C}$.The sequence $(z_n)_{n\ge 0}$ is called the **orbit** of $z_0$ under $f_c$, written

$$
\mathcal{O}_c(z_0)=\{z_0,z_1,z_2,\dots\}.
$$

## **2. Bounded Vs Escaping Orbits**

A central dichotomy in complex dynamics is whether the orbit stays bounded or diverges:

$$
\text{$z_0$ has a bounded orbit} \iff \sup_{n\ge 0}|z_n|<\infty,
$$

$$
\text{$z_0$ escapes to infinity} \iff |z_n|\to\infty \quad (n\to\infty).
$$

The diagram illustrates both behaviors for the **same parameter** $c$ by choosing two different seeds:

 - A seed $z_0=0$ whose orbit remains bounded (blue trajectory),
 - A seed $w_0$ whose orbit eventually escapes (red trajectory).

## **3. The Escape Radius and Why $|z|=2$ Appears**

For quadratic polynomials $f_c(z)=z^2+c$, the circle of radius $2$ plays a fundamental role.

> **Escape radius for $z^2+c$**
> For any $c\in\mathbb{C}$, if at some iterate $n$ one has $|z_n|>2$, then the > orbit diverges:
> $$
> |z_n|>2 \quad\Longrightarrow\quad |z_{n+k}|\to\infty \ \text{as } k\to\infty.
>$$
> **Proof:**
>If $|z|>2$, then
$$
>|f_c(z)|=|z^2+c|\ge |z|^2-|c|.
$$
> Once $|z|$ is sufficiently large, the squaring dominates and $|z|$ grows rapidly.
>In practice, the radius $2$ is a universal threshold used in computations for the quadratic family.


Hence, the dashed circle $|z|=2$ in the plot is an **escape boundary** crossing it certifies divergence. The diagram marks the first iterate $w_{10}$ with $|w_{10}|>2$, after which the red orbit
accelerates outward.

## **4. Julia set and Filled Julia set: What The Picture is Really Probing**

Define the **filled Julia set** of $f_c$ as

$$
K(f_c)=\{z\in\mathbb{C}:\sup_{n\ge 0}|f_c^{\circ n}(z)|<\infty\},
$$

i.e.\ the set of seeds whose orbits are bounded.Its boundary is the **Julia Set:**

$$
J(f_c)=\partial K(f_c).
$$

The diagram is a **local experiment** showing two initial conditions:

$$
z_0\in K(f_c)\quad(\text{blue bounded orbit}), \qquad
w_0\notin K(f_c)\quad(\text{red escaping orbit}).
$$

Thus, the diagram visually separates points that belong to $K(f_c)$ from those that do not.

## **5. Why the Seed $z_0=0$ is Special (Critical Orbit)**

For $f_c(z)=z^2+c$, the unique critical point is $z=0$ (since $f_c'(z)=2z$). The orbit of the critical point is

$$
0,\ c,\ c^2+c,\ (c^2+c)^2+c,\dots
$$

This \emph{critical orbit} governs much of the global dynamics. In particular, the \emph{Mandelbrot set} is defined by

$$
\mathcal{M}=\{c\in\mathbb{C}:\ \sup_{n\ge 0}|f_c^{\circ n}(0)|<\infty\}.
$$

So the blue trajectory (starting at $0$) is simultaneously:

- A bounded orbit in the $z$-plane (dynamics for fixed $c$), and
- Evidence that the parameter $c$ lies in (or near) $\mathcal{M}$.

## **The Mathematics Behind the Diagram (Theory Notes)**

Your code produces four standard pictures that illustrate four key notions in Euclidean spaces $\mathbb{R}^n$. Throughout, we use the Euclidean norm $\|x\|$ and **open balls**

$$
B(x,r)=\{y\in\mathbb{R}^n:\|y-x\|<r\}.
$$

### **(A) Open Set: an Open Disk and the Ball Criterion**

#### **Set drawn in The Code:**

$$
U=\{(x,y)\in\mathbb{R}^2: x^2+y^2<4\}.
$$

#### **Definition (Open Set in $\mathbb{R}^n$):**  
A subset $U\subseteq\mathbb{R}^n$ is **open** if

$$
\forall x\in U,\ \exists r>0 \text{ such that } B(x,r)\subseteq U.
$$

#### **Why the Open Disk is Open:**  
Take any point $x\in U$. Then $\|x\|<2$. Set
$$
r:=2-\|x\|>0.
$$

If $y\in B(x,r)$, then $\|y-x\|<r$, and by the triangle inequality,

$$
\|y\|\le \|y-x\|+\|x\|<r+\|x\|=(2-\|x\|)+\|x\|=2.
$$

Hence $y\in U$. So $B(x,r)\subseteq U$. This is exactly what the inner circle $B(x,r)$ represents in the picture.

>**Boundary Note:** The dashed circle $\|z\|=2$ indicates boundary points are **not included** in an open disk.

### **(B) Disconnected Set: Two Separated Components**

**Set drawn in the code (conceptual form):**  
Two disjoint closed disks, for instance

$$
E=D_1\cup D_2,
\qquad
D_1=\overline{B}((-1.2,0),1),\ \ D_2=\overline{B}((1.2,0),1).
$$

**Definition (Disconnected / Connected).**  
A subset $E\subseteq\mathbb{R}^n$ is **disconnected** if there exist sets $U,V\subseteq E$ such that:
1. $U\neq\varnothing\, V\neq\varnothing$,
2. $U\cap V=\varnothing$,
3. $E=U\cup V$,
4. $U$ and $V$ are **open in the subspace topology on $E$**  
   (equivalently, $U=E\cap G$ and $V=E\cap H$ for some open $G,H\subseteq\mathbb{R}^n$).

**Why this $E$ is Disconnected.**  
Let
$$
G=\{(x,y):x<0\},\qquad H=\{(x,y):x>0\},
$$
both open in $\mathbb{R}^2$. Define
$$
U=E\cap G,\qquad V=E\cap H.
$$
Then $U=D_1, V=D_2$, they are nonempty, disjoint, relatively open in $E$, and $E=U\cup V$. So $E$ is disconnected: it has two separate “pieces”.

### **(C) Path-Connected Set: Dumbbell and Explicit Paths**

#### **Set Drawn in the Code:**  
Two disks joined by a corridor (a “dumbbell” shape). Conceptually:

$$
E=\overline{B}((-a,0),1)\ \cup\ ([-a,a]\times[-\varepsilon,\varepsilon])\ \cup\ \overline{B}((a,0),1),
$$
for some $a>1$, $\varepsilon>0$.

#### **Definition (Path-Connected):**  
A set $E\subseteq\mathbb{R}^n$ is **path-connected** if for every $p,q\in E$ there exists a continuous map

$$
\gamma:[0,1]\to E
\quad\text{such that}\quad
\gamma(0)=p,\ \gamma(1)=q.
$$

#### **Why the Dumbbell is Path-Connected:**  

Given $p,q\in E$:
- Move inside the left disk from $p$ to a point in the corridor,
- Travel along the corridor,
- Move inside the right disk to $q$.


> **Note:** Each part can be taken as a straight line segment (which is continuous). Concatenating finitely many continuous paths gives a continuous path in $E$. So $E$ is path-connected.
>
> **Key Theorem (Always True):**
$$
\text{Path-connected} \ \Longrightarrow\ \text{Connected}.
$$
>So the dumbbell is connected automatically.


### **(D) Region: Open + Connected Example (Open Annulus)**

#### **Set drawn in the code:**

$$
A=\{(x,y)\in\mathbb{R}^2:1<\sqrt{x^2+y^2}<2\}
=\{z\in\mathbb{R}^2:1<\|z\|<2\}.
$$

**Definition (Region in $\mathbb{R}^n$):**  
A **region** is a **nonempty open connected** subset of $\mathbb{R}^n$.

#### **1) Why the Annulus is Open**
Let $p\in A$. Then $1<\|p\|<2$. Define the positive margin

$$
\delta:=\min\{\|p\|-1,\ 2-\|p\|\}>0.
$$

Claim: $B(p,\delta)\subseteq A$.  

If $q\in B(p,\delta)$, then $\|q-p\|<\delta$. By triangle inequality,

$$
\|q\|\ge \|p\|-\|q-p\|>\|p\|-\delta\ge 1,
$$

and also

$$
\|q\|\le \|p\|+\|q-p\|<\|p\|+\delta\le 2.
$$

Hence $1<\|q\|<2$, so $q\in A$. Therefore $A$ is open.

#### **2) Why The Annulus is Connected (Indeed Path-Connected)**
Take any $p,q\in A$. Write in polar form:

$$
p=(r_p\cos\theta_p,\ r_p\sin\theta_p),\qquad
q=(r_q\cos\theta_q,\ r_q\sin\theta_q),
$$

with $1<r_p,r_q<2$. Pick some $r_0\in(1,2)$ (e.g. $r_0=\tfrac{r_p+r_q}{2}$).  
Construct a path inside $A$:

- Radial path from $p$ to $(r_0,\theta_p)$,
- Circular arc at radius $r_0$ from angle $\theta_p$ to $\theta_q$,
- Radial path from $(r_0,\theta_q)$ to $q$.

This stays in $1<r<2$ the whole time, so it gives a continuous $\gamma\subseteq A$. Hence $A$ is path-connected, so connected.

Therefore, $A$ is **open + connected + nonempty**, i.e. a **region**.

## **Connected but Not Path-Connected (Topologist's Sine Curve)**
### **The set:**
Consider the set

$$
S=\Big\{(x,\sin(1/x)):\ 0<x\le 1\Big\}\ \cup\ \Big(\{0\}\times[-1,1]\Big)\ \subset\mathbb{R}^2.
$$

The first part oscillates infinitely often as $x\to 0^+$, and the vertical segment at $x=0$ is included.

### **What the Picture is Conveying?**

The oscillating curve accumulates onto the entire segment $\{0\}\times[-1,1]$:
every point $(0,y)$ with $y\in[-1,1]$ is a limit of points $(x_n,\sin(1/x_n))$ as $x_n\to 0^+$. So the set ``sticks together'' by accumulation, even though it does not provide a traversable corridor for paths.

> ### **Theorem (Classical fact)**
>The set $S$ is connected but not path-connected.
> Proof:
> Let
>
>$$
>C=\{(x,\sin(1/x)):\ 0<x\le 1\}.
>$$
>
>The map $x\mapsto (x,\sin(1/x))$ is continuous on $(0,1]$, and $(0,1]$ is >connected, hence $C$ is connected.
>
>Moreover, the closure $\overline{C}$ is connected (closure of a connected set in $\mathbb{R}^2$ is connected).
One checks that
>$$
\overline{C}=C\cup\big(\{0\}\times[-1,1]\big)=S.
$$
>Therefore $S$ is connected.

Intuitively, any continuous path starting at a point on the vertical segment $\{0\}\times[-1,1]$ would have to move into $x>0$ to reach the oscillating part, but near $x=0$ the curve oscillates infinitely many times between $y=-1$ and $y=1$ at arbitrarily small $x$, creating an obstruction: a path would have to ``choose'' a continuous route through infinitely many oscillations, which is impossible while staying inside $S$.
Formally, one proves that there is no continuous $\gamma:[0,1]\to S$ with $\gamma(0)\in\{0\}\times[-1,1]$ and $\gamma(1)\in C$. Hence $S$ is not path-connected.

## **Visual Proof (Guided by the Diagram)**

Let $U\subseteq \mathbb{R}^n$ be \textbf{open and connected}. Fix a base point $x\in U$ (the point labeled $x$ in the diagram). Define the \emph{reachable set}

$$
A:=\{\,y\in U:\ \text{there exists a continuous path }\gamma:[0,1]\to U
\text{ with }\gamma(0)=x,\ \gamma(1)=y\,\}.
$$

In the picture, $U$ is the **large blue region** (open \& connected), while $A$ is the **orange subregion** inside $U$ consisting of points you can reach from $x$ by moving continuously while staying inside $U$.

### **Step 1: $A\neq\varnothing$**
The constant path $\gamma(t)\equiv x$ lies in $U$ and joins $x$ to itself, so $x\in A$. Hence $A$ is nonempty.

### **Step 2: $A$ is open in $U$ (The Ball Around $y$)**
Take any point $y\in A$ (the point labeled $y\in A$ in the diagram).
Because $U$ is \emph{open}, there exists $r>0$ such that the entire ball

$$
B(y,r)\subset U
$$

(the circle drawn around $y$) lies inside $U$.

Now choose any point $z\in B(y,r)$ (the nearby point labeled $z$). Since Euclidean balls are convex, the straight segment

$$
\sigma(t)=(1-t)y+tz,\qquad t\in[0,1],
$$

stays inside $B(y,r)$, hence inside $U$.

Because $y\in A$, there is already a path $\gamma$ in $U$ from $x$ to $y$
(the thick curved arrow from $x$ into the orange region). Concatenating $\gamma$ with the segment $\sigma$ yields a continuous path in $U$
from $x$ to $z$. Therefore $z\in A$.

Since every $z\in B(y,r)$ lies in $A$, we have

$$
B(y,r)\cap U = B(y,r)\subseteq A,
$$

so $A$ is open in the subspace topology on $U$.

### **Step 3: $U\setminus A$ is open in $U$ (The Ball Around $w$)**

Take a point $w\in U\setminus A$ (the point labeled $w\in U\setminus A$ in the diagram).

Again, openness of $U$ gives an $r>0$ such that

$$
B(w,r)\subset U
$$

(the circle drawn around $w$).

We claim $B(w,r)$ cannot intersect $A$. If it did, pick $z\in B(w,r)\cap A$.
Then there is a path in $U$ from $x$ to $z$ (because $z\in A$). Also, since $z,w\in B(w,r)$, the straight segment from $z$ to $w$ lies entirely inside the ball $B(w,r)\subset U$. Concatenating these two paths produces a path in $U$ from $x$ to $w$, which would imply $w\in A$—a contradiction.

Hence

$$
B(w,r)\cap A=\varnothing
\quad\Longrightarrow\quad
B(w,r)\cap U = B(w,r)\subseteq U\setminus A,
$$

so $U\setminus A$ is open in $U$.

### **Step 4: Connectedness Forces $A=U$.**
We have written $U$ as a disjoint union

$$
U = A\ \cup\ (U\setminus A),
$$

and Steps 2--3 show that both $A$ and $U\setminus A$ are open in $U$.
Since $A\neq\varnothing$ and $U$ is connected, the only possibility is

$$
U\setminus A=\varnothing
\quad\Longrightarrow\quad
A=U.
$$

### **Conclusion (Path-connectedness):}**

Because $A=U$, every point of $U$ can be joined to $x$ by a path in $U$. In particular, given any $p,q\in U$, apply the above argument with $x=p$.
Then $q\in A=U$, so there exists a path in $U$ from $p$ to $q$. Thus $U$ is **path-connected**. $\blacksquare$

## **A Set Can be Open But Not Connected**
Let

$$
E=(-2,-1)\cup(1,2)\subseteq\mathbb{R}.
$$

Decide whether $E$ is a region in $\mathbb{R}$.


**Solution:**

### **Step 1: Check Openness**
Both $(-2,-1)$ and $(1,2)$ are open intervals in $\mathbb{R}$, hence open sets.
A union of open sets is open, so

$$
E=(-2,-1)\cup(1,2)
$$

is open.

### **Step 2: Check Connectedness**
Intuitively $E$ has a gap around $0$. Formally, define

$$
U:=(-2,-1),\qquad V:=(1,2).
$$
Then $U$ and $V$ are nonempty, disjoint, open in $\mathbb{R}$, and
$$
E=U\cup V.
$$
Moreover, $U$ and $V$ are also open in the \emph{subspace topology on $E$}:

$$
U=E\cap(-2,-1),\qquad V=E\cap(1,2),
$$

with $(-2,-1)$ and $(1,2)$ open in $\mathbb{R}$. Thus $E$ is disconnected.

# **More Solved Examples (Proof-based + Numerical Intuition)**

## **Boundary of a Union of Open Intervals in $\mathbb{R}$**
>Let
$$
E=(-2,-1)\cup(1,2)\subset\mathbb{R}.
$$
Compute $\overline{E}$, $E^\circ$, and $\partial E$.

$\textbf{Solution:}$

$\textbf{Step 1: Interior:}$
Both $(-2,-1)$ and $(1,2)$ are open in $\mathbb{R}$, hence every point of each interval is an interior point of $E$.

Therefore

$$
E^\circ=E=(-2,-1)\cup(1,2).
$$

$\textbf{Step 2: Closure:}$ A point $x$ lies in $\overline{E}$ iff every neighborhood of $x$ meets $E$.


- If $x\in(-2,-1)$ or $x\in(1,2)$ then clearly $x\in\overline{E}$.
- For the endpoints: $-2,-1,1,2$ are limit points of $E$ because any interval around each endpoint contains points of the corresponding open interval. Hence
 $$
-2,-1,1,2\in\overline{E}.
 $$
-  If $x\notin[-2,-1]\cup[1,2]$, then the distance from $x$ to $E$ is positive. So we can choose $r>0$ small enough that $(x-r,x+r)$ avoids $E$. Thus $x\notin\overline{E}$. Hence

$$
\overline{E}=[-2,-1]\cup[1,2].
$$


$\textbf{Step 3: Boundary:}$
Using $\partial E=\overline{E}\setminus E^\circ$, we get

$$
\partial E=\bigl([-2,-1]\cup[1,2]\bigr)\setminus\bigl((-2,-1)\cup(1,2)\bigr)=\{-2,-1,1,2\}.
$$

$\textbf{Numerical Visual Check (Neighborhood Test):}$
For $x=-2$ and any $r>0$, the interval $(x-r,x+r)$ contains points in $(-2,-1)\subset E$ and also points $< -2$ which are outside $E$. So $-2\in\partial E$.
For $x=0$ there exists $r<1$ such that $(0-r,0+r)\subset(-1,1)\subset \mathbb{R}\setminus E$, so $0\notin\partial E$.
$\blacksquare$

## **Boundary of $\mathbb{Z}$ in $\mathbb{R}$ (Discrete set)**

Let $A=\mathbb{Z}\subset\mathbb{R}$.
Compute $\overline{\mathbb{Z}}$, $\mathbb{Z}{^\circ}$, and $\partial\mathbb{Z}$.


$\textbf{Solution:}$
$\textbf{Step 1: Interior:}$

Take any $n\in\mathbb{Z}$. For every $r>0$, the neighborhood $(n-r,n+r)$ contains non-integers, hence it is not contained in $\mathbb{Z}$. So $\mathbb{Z}$ has no interior points:

$$
\mathbb{Z}^\circ=\varnothing.
$$

$\textbf{Step 2: Closure:}$
A point $x$ is in $\overline{\mathbb{Z}}$ iff every neighborhood of $x$ meets $\mathbb{Z}$.


- If $x=n\in\mathbb{Z}$, then every neighborhood meets $\mathbb{Z}$ (it contains $n$), so $\mathbb{Z}\subset\overline{\mathbb{Z}}$.

- If $x\notin\mathbb{Z}$, let $d=\min_{n\in\mathbb{Z}}|x-n|>0$. Choose $r=d/2$.
Then $(x-r,x+r)$ contains no integers, so $x\notin\overline{\mathbb{Z}}$.

$$
\overline{\mathbb{Z}}=\mathbb{Z}.
$$


$\textbf{Step 3: Boundary:}$
Using $\partial A=\overline{A}\setminus A^\circ$,

$$
\partial \mathbb{Z}=\overline{\mathbb{Z}}\setminus\mathbb{Z}^\circ=\mathbb{Z}\setminus\varnothing=\mathbb{Z}.
$$

$\textbf{Interpretation:}$ Every integer is a boundary point: every neighborhood hits $\mathbb{Z}$ (contains that integer) and also hits $\mathbb{R}\setminus\mathbb{Z}$ (because it contains non-integers). $\blacksquare$


## **Boundary of a Half-Line**
Let $A=(0,\infty)\subset\mathbb{R}$. Compute $\partial A$.


$\textbf{Solution:}$

We have $\overline{A}=[0,\infty)$ and $A^\circ=A$ because $A$ is open in $\mathbb{R}$.Thus

$$
\partial A=\overline{A}\setminus A^\circ=[0,\infty)\setminus(0,\infty)=\{0\}.
$$

Equivalently, every neighborhood of $0$ contains positive and negative numbers, while points $x>0$ admit a small
neighborhood contained in $A$. $\blacksquare$

## **Numerical Boundary Detector (Algorithmic Viewpoint)**

Let $A\subset\mathbb{R}$ and define a numerical test: for a point $x$, sample decreasing radii $r_k\downarrow 0$ and check whether each neighborhood $(x-r_k,x+r_k)$ contains points in $A$ and in $\mathbb{R}\setminus A$. Apply this to $E=(-2,-1)\cup(1,2)$ at $x=1$ and $x=1.5$.

### **Solution (Proof + Numeric Logic)**

**At $x=1$ (endpoint)**

For any $r>0$, the interval $(1-r,1+r)$ contains points $>1$ and close to $1$, hence points in $(1,2)\subset E$. It also contains points $<1$ (for instance $1-r/2$), which lie outside $E$ because $E$ has a gap $(-1,1)$. So every neighborhood of $1$ meets both $E$ and $\mathbb{R}\setminus E$, hence

$$
1\in\partial E.
$$


**At $x=1.5$ (Interior Point):**
Choose $r=0.4$. Then $(1.5-0.4,1.5+0.4)=(1.1,1.9)\subset(1,2)\subset E$.
So $x=1.5$ has a neighborhood contained in $E$, hence

$$
1.5\in E^\circ \quad\Rightarrow\quad 1.5\notin\partial E.

```python
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# Helper routines (numerical "boundary test" intuition)
# -----------------------------
def interval_hits(A_indicator, x, r, sample_points=800):
    """
    Numerical check: does (x-r, x+r) contain points from A and from complement?
    A_indicator: function mapping float -> bool (True if in A)
    """
    xs = np.linspace(x - r, x + r, sample_points)
    inA = np.array([A_indicator(t) for t in xs], dtype=bool)
    return inA.any(), (~inA).any()

def boundary_test(A_indicator, x, radii, sample_points=800):
    """
    Returns a list of (r, hitA, hitComp) and a boolean 'looks_boundary'
    """
    results = []
    looks_boundary = True
    for r in radii:
        hitA, hitComp = interval_hits(A_indicator, x, r, sample_points=sample_points)
        results.append((r, hitA, hitComp))
        if not (hitA and hitComp):
            looks_boundary = False
    return results, looks_boundary

# -----------------------------
# Example 1: A = Z inside R
# Membership: "is integer" with a tolerance (floating sampling)
# -----------------------------
def in_Z(t, tol=1e-9):
    return abs(t - np.round(t)) < tol

# Radii for boundary intuition
radii = [1.0, 0.4, 0.2, 0.08, 0.03]

# Points to test: integer and non-integer
test_points_Z = [2.0, 2.3]

results_Z = {}
for x0 in test_points_Z:
    res, looks = boundary_test(in_Z, x0, radii, sample_points=1200)
    results_Z[x0] = (res, looks)

# -----------------------------
# Example 2: A = (0,1) union (2,3)
# Membership is exact from inequalities
# -----------------------------
def in_A_union(t):
    return (0.0 < t < 1.0) or (2.0 < t < 3.0)

test_points_union = [1.0, 0.5, 1.2, 2.0, 2.5]  # boundary, interior, outside, boundary, interior
results_union = {}
for x0 in test_points_union:
    res, looks = boundary_test(in_A_union, x0, radii, sample_points=1200)
    results_union[x0] = (res, looks)

# -----------------------------
# Visualizations
# -----------------------------
# Plot 1: Z in R and intervals around points
plt.figure(figsize=(10, 3.3))
ax1 = plt.gca()
ax1.set_title("Numerical boundary intuition: A = Z in R (interval neighborhoods)")
ax1.set_xlabel("x")
ax1.set_yticks([])
ax1.set_ylim(-1, 1)

# Plot integers in a range
ints = np.arange(-2, 7)
ax1.scatter(ints, np.zeros_like(ints), marker='|', s=400, linewidths=3)

# Two test points
ax1.scatter([2.0], [0.3], s=40)
ax1.text(2.0, 0.45, "x = 2 (integer)", ha='center')
ax1.scatter([2.3], [-0.3], s=40)
ax1.text(2.3, -0.45, "x = 2.3 (non-integer)", ha='center')

# Draw a couple of neighborhoods
for (x0, y) in [(2.0, 0.3), (2.3, -0.3)]:
    for r in [0.2, 0.08]:
        ax1.plot([x0 - r, x0 + r], [y, y], linewidth=6, alpha=0.35)

ax1.grid(True, axis='x', alpha=0.25)
plt.tight_layout()
plt.show()

# Plot 2: Union of intervals in R and boundary points
plt.figure(figsize=(10, 3.3))
ax2 = plt.gca()
ax2.set_title("Numerical boundary intuition: A = (0,1) ∪ (2,3) in R")
ax2.set_xlabel("x")
ax2.set_yticks([])
ax2.set_ylim(-1, 1)

# Draw the set A as thick segments (open intervals)
ax2.plot([0, 1], [0, 0], linewidth=10, alpha=0.25)
ax2.plot([2, 3], [0, 0], linewidth=10, alpha=0.25)
ax2.text(0.5, 0.25, "(0,1)", ha='center')
ax2.text(2.5, 0.25, "(2,3)", ha='center')

# Mark boundary points 0,1,2,3
for b in [0, 1, 2, 3]:
    ax2.scatter([b], [0], s=40)
    ax2.text(b, -0.25, f"{b}", ha='center')

# Sample test points and one neighborhood radius
for x0, y in [(1.0, 0.55), (0.5, 0.35), (1.2, 0.15), (2.0, -0.15), (2.5, -0.35)]:
    ax2.scatter([x0], [y], s=40)
    ax2.text(x0, y + 0.12, f"x={x0}", ha='center', fontsize=9)
    r = 0.2
    ax2.plot([x0 - r, x0 + r], [y, y], linewidth=6, alpha=0.35)

ax2.grid(True, axis='x', alpha=0.25)
plt.tight_layout()
plt.show()

# -----------------------------
from IPython.display import display, Markdown

def md_table_from_results(results_dict, title=None, digits_r=2):
    """
    results_dict format:
      { x0: (rows, looks) }
    where rows is list of tuples: (r, hitA, hitC)
    """
    lines = []

    if title:
        lines.append(f"### {title}\n")

    # Header
    lines.append("| x | r | hits A? | hits ℝ \\ A? |")
    lines.append("|---:|---:|:------:|:-----------:|")

    # Stable ordering
    for x0 in sorted(results_dict.keys()):
        rows, _looks = results_dict[x0]

        # For readability: keep x shown once per block
        first = True
        for (r, hitA, hitC) in rows:
            x_cell = f"{x0}" if first else ""
            first = False

            r_cell = f"{r:.{digits_r}g}"  # compact numeric formatting
            hitA_cell = "✅" if hitA else "❌"
            hitC_cell = "✅" if hitC else "❌"

            lines.append(f"| {x_cell} | {r_cell} | {hitA_cell} | {hitC_cell} |")

        # separator row between x-blocks
        lines.append("|  |  |  |  |")

    return "\n".join(lines)

# ---- build Markdown tables ----
md_Z = md_table_from_results(results_Z, title="Table for A = ℤ (example)")
md_union = md_table_from_results(results_union, title="Table for A = (0,1) ∪ (1,2) (example)")

# ---- display ----
display(Markdown(md_Z))
display(Markdown(md_union))

```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# ----------------------------
# Geometry helpers
# ----------------------------
def norm(v):
    return float(np.sqrt(v[0]**2 + v[1]**2))

def in_B(pt):  # unit open disk: x^2+y^2 < 1
    return norm(pt) < 1.0

def numeric_boundary_test_for_disk(p, radii):
    """
    For each radius r, we test whether B(p,r) contains:
      - a point inside B (unit disk)
      - a point outside B
    We do this using the 'radial in/out' construction:
        p_- = (1 - r/4) p,   p_+ = (1 + r/4) p.
    """
    rows = []
    p = np.array(p, dtype=float)
    for r in radii:
        pm = (1 - r/4) * p
        pp = (1 + r/4) * p

        # Check membership
        hit_inside = in_B(pm)
        hit_outside = (not in_B(pp))

        # Check both points lie within the ball centered at p of radius r
        dist_pm = norm(pm - p)
        dist_pp = norm(pp - p)

        rows.append({
            "r": r,
            "p_- in B?": hit_inside,
            "p_+ outside B?": hit_outside,
            "||p_- - p|| < r": dist_pm < r,
            "||p_+ - p|| < r": dist_pp < r,
            "dist(p_-,p)": dist_pm,
            "dist(p_+,p)": dist_pp,
        })
    return pd.DataFrame(rows)

# ----------------------------
# Pick a boundary point p on the unit circle
# ----------------------------
theta = np.deg2rad(35)
p = np.array([np.cos(theta), np.sin(theta)])

# radii to test (decreasing)
radii = [0.6, 0.3, 0.15, 0.08, 0.04]

df = numeric_boundary_test_for_disk(p, radii)
df

r_plot = 0.22  # choose one radius to visualize

pm = (1 - r_plot/4) * p
pp = (1 + r_plot/4) * p

fig, ax = plt.subplots(figsize=(6,6))

# Unit disk (filled)
t = np.linspace(0, 2*np.pi, 400)
ax.fill(np.cos(t), np.sin(t), alpha=0.15)  # no explicit color

# Boundary circle
ax.plot(np.cos(t), np.sin(t), linewidth=2)

# Ball around p of radius r_plot
ball_x = p[0] + r_plot*np.cos(t)
ball_y = p[1] + r_plot*np.sin(t)
ax.plot(ball_x, ball_y, linewidth=2)
ax.fill(ball_x, ball_y, alpha=0.10)

# Points: p, p_-, p_+
ax.scatter([p[0]],  [p[1]],  s=60)
ax.scatter([pm[0]], [pm[1]], s=60)
ax.scatter([pp[0]], [pp[1]], s=60)

# Annotate
ax.annotate(r"$p\in S$", xy=p, xytext=(p[0]+0.18, p[1]+0.10),
            arrowprops=dict(arrowstyle="->", lw=1.5),
            fontsize=12)

ax.annotate(r"$p_- \in B$", xy=pm, xytext=(pm[0]-0.65, pm[1]+0.10),
            arrowprops=dict(arrowstyle="->", lw=1.5),
            fontsize=12)

ax.annotate(r"$p_+ \notin B$", xy=pp, xytext=(pp[0]+0.10, pp[1]-0.35),
            arrowprops=dict(arrowstyle="->", lw=1.5),
            fontsize=12)

ax.annotate(r"$B(p,r)$", xy=(p[0], p[1]+r_plot), xytext=(p[0]-0.05, p[1]+r_plot+0.15),
            fontsize=12)

ax.set_aspect("equal", adjustable="box")
ax.set_xlim(-1.4, 1.6)
ax.set_ylim(-1.4, 1.6)
ax.set_title("Numeric Boundary Test Picture for The Unit Disk")
ax.grid(True, alpha=0.25)

plt.show()
```
